```python
import pandas as pd

file_name = next(iter(uploaded))
df = pd.read_excel(file_name)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

display(df.info())
```

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Remove extra spaces from column names
df_clean.columns = df_clean.columns.str.strip()

# Check missing values
print("Missing values in each column:")
print(df_clean.isnull().sum())

print("\nOriginal dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)

display(df_clean.head())
```

```python
output_file = "RetailPulse_Cleaned.xlsx"

df_clean.to_excel(output_file, index=False)

print("Cleaned file saved successfully!")
```

```python
# RetailPulse: Basic Data Analysis

print("Dataset Shape (Rows, Columns):")
print(df_clean.shape)

print("\nFirst 5 Rows:")
display(df_clean.head())

print("\nDataset Information:")
df_clean.info()

print("\nMissing Values in Each Column:")
print(df_clean.isnull().sum())
```

```python
# RetailPulse - Overall Business Performance

total_sales = df_clean["Sum of Total Sales"].sum()
total_cost = df_clean["Sum of Estimated Cost"].sum()
total_profit = df_clean["Sum of Estimated Profit"].sum()

print("===== RetailPulse Business Summary =====")
print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Estimated Cost: {total_cost:,.2f}")
print(f"Total Estimated Profit: {total_profit:,.2f}")

if total_sales != 0:
    profit_margin = (total_profit / total_sales) * 100
    print(f"Profit Margin: {profit_margin:.2f}%")
else:
    print("Profit Margin: Cannot calculate because sales are zero.")
```

```python
# RetailPulse - Country-wise Sales Analysis

country_sales = (
    df_clean.groupby("Country")["Sum of Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Countries by Total Sales:")
display(country_sales.head(10).to_frame())

print("\nTop Country by Sales:")
print(country_sales.index[0])

print("\nHighest Country Sales:")
print(round(country_sales.iloc[0], 2))
```

```python
import matplotlib.pyplot as plt

# Select top 10 countries
top_countries = country_sales.head(10).sort_values()

# Create bar chart
plt.figure(figsize=(10, 6))
top_countries.plot(kind="barh")

plt.title("RetailPulse - Top 10 Countries by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Country")
plt.tight_layout()

plt.show()
```

```python
# RetailPulse - Product-wise Sales Analysis

product_sales = (
    df_clean.groupby("Description")["Sum of Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products by Sales:")
display(product_sales.head(10).to_frame())

print("\nTop Selling Product:")
print(product_sales.index[0])

print("\nSales of Top Selling Product:")
print(round(product_sales.iloc[0], 2))
```

```python
# RetailPulse - Monthly Sales Analysis

monthly_sales = (
    df_clean.groupby("Year Month")["Sum of Total Sales"]
    .sum()
    .sort_index()
)

print("Monthly Sales:")
display(monthly_sales.to_frame())

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))
monthly_sales.plot(kind="line", marker="o")

plt.title("RetailPulse - Monthly Sales Trend")
plt.xlabel("Year Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# RetailPulse - Quantity Analysis

product_quantity = (
    df_clean.groupby("Description")["Sum of Quantity"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products by Quantity Sold:")
display(product_quantity.head(10).to_frame())
```

```python
import matplotlib.pyplot as plt

# Calculate business metrics
metrics = {
    "Total Sales": df_clean["Sum of Total Sales"].sum(),
    "Estimated Cost": df_clean["Sum of Estimated Cost"].sum(),
    "Estimated Profit": df_clean["Sum of Estimated Profit"].sum()
}

plt.figure(figsize=(9, 5))
plt.bar(metrics.keys(), metrics.values())

plt.title("RetailPulse - Sales, Cost and Profit")
plt.ylabel("Amount")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

print("Business Metrics:")
for name, value in metrics.items():
    print(f"{name}: {value:,.2f}")
```

```python
# RetailPulse - Customer Segment Analysis

segment_sales = (
    df_clean.groupby("Customer Segment")["Sum of Total Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("Sales by Customer Segment:")
display(segment_sales.to_frame())

plt.figure(figsize=(8, 5))
segment_sales.plot(kind="bar")

plt.title("RetailPulse - Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```
